```python
def calc_QliqVogel_m3Day(P_test, Pr, Pb, pi, Wc):
    """
    Calculates the liquid flow rate at a given test pressure using the Vogel method.

    Args:
        P_test (float): Test pressure, atm
        Pr (float): Reservoir pressure, atm
        Pb (float): Bubble point pressure, atm
        pi (float): Productivity index, m3/day/atm
        Wc (float): Water cut, %

    Returns:
        float: Liquid flow rate at the test pressure, m3/day
    """

    if P_test < 0:
        raise ValueError("P_test cannot be negative")
    if Pr < 0:
        raise ValueError("Pr cannot be negative")
    if Pb < 0:
        raise ValueError("Pb cannot be negative")
    if pi < 0:
        raise ValueError("pi cannot be negative")

    if Pr < Pb:
        Pb = Pr

    qb = pi * (Pr - Pb)

    if Wc > 100:
        Wc = 100
    elif Wc < 0:
        Wc = 0

    if Wc == 100 or P_test >= Pb:
        return pi * (Pr - P_test)
    else:
        fw = Wc / 100
        fo = 1 - fw
        qo_max = qb + (pi * Pb) / 1.8
        p_wfg = fw * (Pr - qo_max / pi)

        if P_test > p_wfg:
            a = 1 + (P_test - (fw * Pr)) / (0.125 * fo * Pb)
            B = fw / (0.125 * fo * Pb * pi)
            c = (2 * a * B) + 80 / (qo_max - qb)
            d = (a ** 2) - (80 * qb / (qo_max - qb)) - 81
            if B == 0:
                return abs(d / c)
            else:
                return (-c + ((c ** 2 - 4 * B ** 2 * d) ** 0.5)) / (2 * B ** 2)
        else:
            CG = 0.001 * qo_max
            cd = fw * (CG / pi) + fo * 0.125 * Pb * (-1 + (1 + 80 * ((0.001 * qo_max) / (qo_max - qb))) ** 0.5)
            return (p_wfg - P_test) / (cd / CG) + qo_max
```

```python
import numpy as np

def calc_QliqVogel_m3Day(P_test, Pr, Pb, pi, Wc):
    """
    Calculates the liquid flow rate at a given test pressure using the Vogel method.

    Args:
        P_test (float or array): Test pressure, atm
        Pr (float or array): Reservoir pressure, atm
        Pb (float or array): Bubble point pressure, atm
        pi (float or array): Productivity index, m3/day/atm
        Wc (float or array): Water cut, %

    Returns:
        float or array: Liquid flow rate at the test pressure, m3/day
    """

    # Ensure inputs are NumPy arrays
    P_test, Pr, Pb, pi, Wc = (np.asarray(x) for x in (P_test, Pr, Pb, pi, Wc))

    # Clip inputs to valid ranges
    P_test, Pr, Pb, pi = (np.clip(x, 0, None) for x in (P_test, Pr, Pb, pi))
    Wc = np.clip(Wc, 0, 100)

    # Calculate qb
    qb = pi * (Pr - Pb)

    # Calculate qo_max
    qo_max = qb + (pi * Pb) / 1.8

    # Calculate p_wfg
    fw = Wc / 100
    fo = 1 - fw
    p_wfg = fw * (Pr - qo_max / pi)

    # Calculate QliqVogel
    mask1 = (Wc == 100) | (P_test >= Pb)
    mask2 = ~mask1 & (P_test > p_wfg)
    mask3 = ~mask1 & ~mask2

    QliqVogel = np.zeros_like(P_test)
    QliqVogel[mask1] = pi[mask1] * (Pr[mask1] - P_test[mask1])

    a = 1 + (P_test[mask2] - (fw[mask2] * Pr[mask2])) / (0.125 * fo[mask2] * Pb[mask2])
    B = fw[mask2] / (0.125 * fo[mask2] * Pb[mask2] * pi[mask2])
    c = (2 * a * B) + 80 / (qo_max[mask2] - qb[mask2])
    d = (a ** 2) - (80 * qb[mask2] / (qo_max[mask2] - qb[mask2])) - 81
    QliqVogel[mask2] = (-c + np.sqrt(c ** 2 - 4 * B ** 2 * d)) / (2 * B ** 2)

    CG = 0.001 * qo_max[mask3]
    cd = fw[mask3] * (CG / pi[mask3]) + fo[mask3] * 0.125 * Pb[mask3] * (-1 + np.sqrt(1 + 80 * ((0.001 * qo_max[mask3]) / (qo_max[mask3] - qb[mask3]))))
    QliqVogel[mask3] = (p_wfg[mask3] - P_test[mask3]) / (cd / CG) + qo_max[mask3]

    return QliqVogel
```
